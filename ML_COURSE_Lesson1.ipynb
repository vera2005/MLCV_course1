```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from skimage import data
```

**1. Загрузка изображения**
Загружаем встроенное изображение и выводим на экран

Варианты: astronaut, cat, coffee, retina, ограничимся)

```python
img_rgb = data.cat()  # Это RGB-изображение 512x512x3
# Показываем изображение
plt.figure(figsize=(6,6))  # 1. Создаем квадратный холст 6x6 дюймов
plt.imshow(img_rgb)        # 2. Рисуем на нем наше изображение
plt.title("Оригинал")      # 3. Добавляем заголовок
plt.axis('off')            # 4. Убираем оси координат
plt.show()                 # 5. Показываем результат
```

```python
print(f"Размер изображения: {img_rgb.shape}")  # (512, 512, 3)
print(f"Тип данных: {img_rgb.dtype}")          # uint8
print(f"Минимальное значение пикселя: {img_rgb.min()}, максимальное: {img_rgb.max()}")
```

**2.Изображение как матрица чисел**


```python
print("Первые 2x1 пикселей (верхний левый угол):")
print(img_rgb[:2, :1])   # форма (2,1,3) — каждый пиксель — три числа (R,G,B)
```

```python
# Создаём функцию для отображения нескольких изображений в ряд
def show_images(images, titles, cols=3):
    fig, axes = plt.subplots(1, cols, figsize=(15, 5))
    for i, (img, title) in enumerate(zip(images, titles)):
        if len(img.shape) == 2:
            axes[i].imshow(img, cmap='gray')
        else:
            axes[i].imshow(img)
        axes[i].set_title(title)
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()
```

**3. Базовые геометрические преобразования**


```python
# 3.1. Отражение по горизонтали и вертикали
flip_h = cv2.flip(img_rgb, 1)   # 1 — по горизонтали
flip_v = cv2.flip(img_rgb, 0)   # 0 — по вертикали

show_images([img_rgb, flip_h, flip_v],
            ["Оригинал", "Отражение по горизонтали", "Отражение по вертикали"], cols=3)
```

```python
# 3.2. Поворот на 90°, 180°, 270°
rot_90 = cv2.rotate(img_rgb, cv2.ROTATE_90_CLOCKWISE)
rot_180 = cv2.rotate(img_rgb, cv2.ROTATE_180)
rot_270 = cv2.rotate(img_rgb, cv2.ROTATE_90_COUNTERCLOCKWISE)

show_images([rot_90, rot_180, rot_270],
            ["Поворот 90°", "Поворот 180°", "Поворот 270°"], cols=3)
```

4. Изменение размера (масштабирование)


```python
# Увеличение изображения в 2 раза
height, width = img_rgb.shape[:2]
resized_large = cv2.resize(img_rgb, (width*2, height*2), interpolation=cv2.INTER_CUBIC)

# Показываем только увеличенное изображение
plt.figure(figsize=(10, 10))
plt.imshow(resized_large)
plt.title(f"Увеличенное в 2 раза (размер: {resized_large.shape[1]}x{resized_large.shape[0]})")
plt.axis('off')
plt.show()
```

5. Обрезка (crop)


```python
# Вырежем центральную часть размером 200x200 пикселей
x1 = width//2 - 100
x2 = width//2 + 100
y1 = height//2 - 100
y2 = height//2 + 100
cropped = img_rgb[y1:y2, x1:x2]

plt.figure(figsize=(4,4))
plt.imshow(cropped)
plt.title("Обрезка центра 200x200")
plt.axis('off')
plt.show()
```

6. Работа с цветовыми пространствами

Важно: OpenCV использует порядок BGR, а skimage — RGB. Поэтому при переводе через OpenCV нужно сначала преобразовать.



```python
# Преобразуем в BGR (как в OpenCV)
img_bgr = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)

# ОШИБКА: показываем BGR как RGB (цвета искажены)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Правильно: RGB (оригинал)")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_bgr)   # неверно: BGR интерпретируется как RGB
plt.title("ОШИБКА: BGR показан как RGB\n(синий и красный перепутаны)")
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Правильный перевод в Grayscale из RGB (без лишней конвертации)
img_gray_correct = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

plt.figure(figsize=(15, 5))
plt.imshow(img_gray_correct, cmap='gray')
plt.title("Правильно: RGB -> Grayscale")
plt.axis('off')
plt.show()
```

# Нейрон

1. Функции активации

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)
```

2. Класс нейрона

```python
class Neuron:
    def __init__(self, weights, bias, activation='sigmoid'):
        self.weights = np.array(weights)
        self.bias = bias
        self.activation = sigmoid if activation == 'sigmoid' else tanh if activation == 'tanh' else relu

    def forward(self, inputs):
        z = np.dot(inputs, self.weights) + self.bias
        return self.activation(z)

    def __call__(self, inputs):
        return self.forward(inputs)
```

3. Создание нейрона с одним входом

```python
neuron1 = Neuron(weights=[0.5], bias=-0.2, activation='sigmoid')
print("--- Нейрон с одним входом ---")
for x in [-2, -1, 0, 1, 2]:
    out = neuron1.forward([x])
    print(f"x = {x:2} -> выход = {out:.4f}")
```

4. Построение графика

```python
x_vals = np.linspace(-5, 5, 100)
y_vals = [neuron1.forward([x]) for x in x_vals]

plt.figure(figsize=(6,4))
plt.plot(x_vals, y_vals, linewidth=2)
plt.axhline(0.5, color='gray', linestyle='--', label='порог 0.5')
plt.xlabel('Вход x')
plt.ylabel('Выход нейрона')
plt.title('Сигмоидный нейрон с w=0.5, b=-0.2')
plt.grid(True)
plt.legend()
plt.show()
```

Нейрон с двумя входами - линейный классификатор

```python
import numpy as np
import matplotlib.pyplot as plt


# Генерируем два класса точек (линейно разделимые)
np.random.seed(42)
class0 = np.random.randn(20, 2) * 0.5 - [0.5, 0.5]   # центр в (-0.5, -0.5)
class1 = np.random.randn(20, 2) * 0.5 + [1.5, 1.5]   # центр в (1.5, 1.5)

# Рисуем точки
plt.figure(figsize=(6,5))
plt.scatter(class0[:,0], class0[:,1], color='red', label='Класс 0')
plt.scatter(class1[:,0], class1[:,1], color='blue', label='Класс 1')



plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Нейрон разделяет два класса')
plt.legend()
plt.grid(alpha=0.3)
plt.axis('equal')  # чтобы масштаб по осям был одинаковым
plt.show()
```

Задача: подобрать такие значения весов и смещения, чтобы получить корректную границу решения (разделяющую прямую)

```python
# Параметры нейрона
w1, w2 = 1.0, 1.0
b = -1.5

plt.figure(figsize=(6,5))
plt.scatter(class0[:,0], class0[:,1], color='red', label='Класс 0')
plt.scatter(class1[:,0], class1[:,1], color='blue', label='Класс 1')

# Рисуем разделяющую прямую: w1*x1 + w2*x2 + b = 0 → x2 = -(w1/w2)*x1 - b/w2
x_line = np.linspace(-2, 3, 100)
y_line = -(w1/w2) * x_line - b/w2
plt.plot(x_line, y_line, 'k--', linewidth=2, label='Граница решения')

plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Нейрон разделяет два класса')
plt.legend()
plt.grid(alpha=0.3)
plt.axis('equal')  # чтобы масштаб по осям был одинаковым
plt.show()
```
