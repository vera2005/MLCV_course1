```python
# Создаём функцию для отображения нескольких изображений в ряд
def show_images(images, titles, cols=3):
    fig, axes = plt.subplots(1, cols, figsize=(15, 5))
    for i, (img, title) in enumerate(zip(images, titles)):
        if len(img.shape) == 2:
            axes[i].imshow(img, cmap='gray')
        else:
            axes[i].imshow(img)
        axes[i].set_title(title)
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()
```

**3. Базовые геометрические преобразования**


```python
# 3.1. Отражение по горизонтали и вертикали
flip_h = cv2.flip(img_rgb, 1)   # 1 — по горизонтали
flip_v = cv2.flip(img_rgb, 0)   # 0 — по вертикали

show_images([img_rgb, flip_h, flip_v],
            ["Оригинал", "Отражение по горизонтали", "Отражение по вертикали"], cols=3)
```

```python
# 3.2. Поворот на 90°, 180°, 270°
rot_90 = cv2.rotate(img_rgb, cv2.ROTATE_90_CLOCKWISE)
rot_180 = cv2.rotate(img_rgb, cv2.ROTATE_180)
rot_270 = cv2.rotate(img_rgb, cv2.ROTATE_90_COUNTERCLOCKWISE)

show_images([rot_90, rot_180, rot_270],
            ["Поворот 90°", "Поворот 180°", "Поворот 270°"], cols=3)
```
